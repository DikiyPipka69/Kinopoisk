Создадим из этого таблицу, предварительно очистив строки от лишней информации

```python
import re
s = '<текст>8.3<текст>'
PATTERN = re.compile(r'[>].+[<]', re.IGNORECASE)
re.search(PATTERN, s)
```

```python
def get_value(s):
    '''на входе информация с тегами'''
    s = str(s)
    ss = ''
    PATTERN = re.compile(r'[>].+[<]', re.IGNORECASE)
    match = re.search(PATTERN, s)
    if match:
        s = match.group(0)
        ss = s.replace('<','').replace('>','')
    else:
        ss = ''
    return ss
```

```python
get_value('<span class="rating__value rating__value_positive" data-tid="59e7b831">8.3</span>')
```

```python
import pandas as pd
res = []

names = soup.find_all('p', {'class': 'selection-film-item-meta__name'})
names = [get_value(i) for i in names]

rating = soup.find_all('span', {'class': 'rating__value rating__value_positive'})
rating = [get_value(i) for i in rating]

meta = soup.find_all('span', {'class': 'selection-film-item-meta__meta-additional-item'})
meta = [get_value(i) for i in meta]
country = meta[0::2]
genre = meta[1::2]


films_data = {
    'Название' : names,
    'Рэйтинг' : rating,
    'Страна' : country,
    'Жанр' : genre
}


# создаем датафрейм
df = pd.DataFrame(films_data)
```
